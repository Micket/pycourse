## Load text file to list of lines

To read the file `ca1_step1_input_data.txt` use the `open` function together with a `widh` statement to avoid having to explicitly close the file. Read every line of the file into a list of lines (let's call it `lines`) using the method `.readlines()`.

````python
# -- Solution

with open('ca1_step1_input_data.txt', 'r') as fd:
    lines = fd.readlines()
````

## Parse main header

To parse the main header of the file (on line one and two) use the string methods `.split`, `.replace`, etc.. To convert from a string to a number use the `int(...)` and `float(...)` built in type functions.
- <https://docs.python.org/3.7/library/stdtypes.html#text-sequence-type-str>
- <https://docs.python.org/3.7/library/stdtypes.html#numeric-types-int-float-complex>

Store the numbers in variables with the same names as liste in the first header line, i.e. `time_steps`, `time_step`, `radius`, `v_variance`, and `N_particles`.

````python
# -- Solution

def parse_line(line):
    return [float(value) for value in line.replace(';', ',').split(',')]

header, lines = lines[1], lines[2:]
time_steps, time_step, radius, v_variance, N_particles = parse_line(header)
time_steps, N_particles = int(time_steps), int(N_particles)
````

When you have parsed the data you should be able to print all the variables from the header. The result should look like this:

````python
print(time_steps, time_step, radius, v_variance, N_particles)
````

## Locate the start of each time-step block

To parse the rest of the file let's locate the start of each time step by looking for the time-step header `# time_step` and then parse the `N_particles` rows of position and velocity data in each time-step block.

To search for the header you can use the `.find` string method or the python built in `in` operator:

````python
'Py' in 'Python'
````

Loop over the `lines` and store the index of the line where each time step block starts in a list called `block_line_idxs`. Make sure to use pythonic looping and the `enumerate` function to obtain the line index. 

````python
# -- Solution

block_line_idxs = [idx for idx, line in enumerate(lines) if 'time_step' in line]
````

The beginning of `block_line_idxs` should look something like:

````python
print('block_line_idxs =', block_line_idxs[:8])
````

## Parse positions and velocities for all particles in each time step block

Now you are ready to parse the data in each time-step block! Store the data in a (three level) nested list called `data`. 
 1. The first level should be the time step index, 
 2. the second level the particle index, 
 3. and the third level a list of the four values of the `x`, `y` position- and `v_x`, `v_y` velocitiy-components stored as floating point numbers (`float`).

Note that the spacings in the lines with position and velocity data are a bit random so it is important to construct a robust parser.

````python
# -- Solution

blocks_lines = [lines[idx + 2:idx + 2 + N_particles] for idx in block_line_idxs]
data = [[parse_line(line) for line in block_lines] for block_lines in blocks_lines]
````

To check the dimensions and type of the elements of nested list `data` you can use the `len` and `type` functions. Printing the info on the three levels and the type should look like:

````python
print(len(data), len(data[0]), len(data[0][0]), type(data[0][0][0]))
````

## Convert a nested list to a multidimensional (NumPy) array

While it is possible to work solely with nested lists, using the multidimensional array type `ndarray` from NumPy is much more convenient. Therefore convert the nested list to a three dimensional array using the `array` function in the `numpy` module.

The established coding style for using `numpy` is to import it using
```python
import numpy as np
```
and then access its functions by writing, e.g. `np.arrray`. The function `np.array` can directlt convert from nested lists to `ndarray` so use it to re-define the `data` variable as a `ndarray`. 

````python
# -- Solution

import numpy as np
data = np.array(data)
````

After this you should be able to look at the type, the element type (also called the `dtype`), and shape of `data`

## Split position vectors and velocity vectors

To make the numerical analysis easier you want to store the final result in two separate `numpy` `ndarrays`, one for the positions called `R` and one for the velocities called `V`. Use the NumPy slicing syntax to split `data` in `R` and `V`, see
- <https://docs.scipy.org/doc/numpy/reference/arrays.indexing.html>

In order to make statistical analysis simpler we also want to change the order of the indices, keeping the first index as the time step, but letting the second index being the cartesian direction (i.e. `x`  and `y`), and the third the particle index.

To swap the order of the axes the NumPy "transpose-like" manipulation routines are useful.
- <https://docs.scipy.org/doc/numpy/reference/routines.array-manipulation.html#transpose-like-operations>

````python
# -- Solution

R = np.swapaxes(data[..., :2], 1, 2)
V = np.swapaxes(data[..., 2:], 1, 2)
````

## Storing the Python objects using the `pickle` module

Before starting to postprocess the data we want to store the position and velocity arrays (`R` and `V`) and the header data (`radius`, `time_step`, and `v_variance`) and to file. To this end you will use the `pickle` module that can write (nearly) any Python object to file, see <https://docs.python.org/3/library/pickle.html#examples>.

Make tuple `t` of the data, i.e. `t = (R, V, radius, time_step, v_variance)` and "pickle" it to a file with the name `ca1_step1_output_data.pkl`. Note, *please store the objects in the same order as in the tuple.* You are going to be using this file in the second part of this assignment.

The `pkl` file format is a so called "binary" file format which stores its contents as binary data. This is much more space efficient than storing numerical data in text files, as you can see by comparing the size of the input file `ca1_step1_input_data.txt` (29 Mb) and your final `ca1_step1_output_data.pkl` file (12 Mb).

````python
# -- Solution

import pickle
with open('ca1_step1_output_data.pkl', 'wb') as fd:
    pickle.dump((R, V, radius, time_step, v_variance), fd)
````
