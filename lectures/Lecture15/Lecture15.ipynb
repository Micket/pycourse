# Functions and input/output


Know the difference between a function that
- **modifies** the input data, and one that
- **returns a modified copy**. 

Examples

````python
def increment(values):
    values[-1] = 0
````

````python
def increment(values):
    new_values = values.copy()
    new_values[-1] = 0
    return new_values
````

````python
x = [1,2,3]
y = increment(x)
print(y, x)
````

````python
def increment(values):
    new_values = values + [4,5]
    return new_values
````

# `tuple`, `list`, `set`, and `dict`

What is a
- `tuple`, `()`
- `list`, `[]`
- `set`, `{}`
- `dict`, `{}`

````python
t = tuple(a**2 for a in range(-5, 5))
print(t)
````

````python
l = [a**2 for a in range(-5, 5)]
print(l)
````

````python
s = {a**2 for a in range(-5,5)}
print(s)
````

````python
d = {a : a**2 for a in range(-5, 5)}
print(d)
d[-3] = 8
print(d)
for value in d.values():
    print(value)
````

# Immutability

## Challenge
Why *is* `x += (4, 5)` the same as `x = x + (4, 5)` ?

- Some Python types are immutable
- In particular **strings** and **tuples** (though, the object they point at can change internally) <br>
- To modify them, you have to copy.

````python
str, tuple, int, float, bool
````

````python
x = 1
y = x
x += 1
print(y)
````

````python
t = (1, 2, 3)
print(t[0])
t[0] = 3
print(t[0])
````

````python
t = (1, [2,3,4], 5)
t[1][1] = 6
print(t)
````

````python
x = 'Hello '
y = x
x += 'World'
````
