# Assignment, references and inplace operations

## Challenge
Why is `x += [4, 5]` not the same as `x = x + [4, 5]`

````python
x = [1,2,3]
y = x
x += [4, 5]
print(y)
````

````python
x = [1,2,3]
y = x
x = x + [4, 5]
print(y)
````

- Assignment, i.e. `=` only passes references
- Standard operators generate a new object, `a + b`
- Inplace operators modify the current object, `a += b`

# Functions and input/output

Know the difference between a function that
- **modifies** the input data, and one that
- **returns a modified copy**. 

For example, a method that modifies values in place might look like:

````python
def increment(values):
    values += 1
````

````python
def increment(values):
    return values + 1
````

````python
def increment(values):
    new_values = values + 1
    return new_values
````

# `tuple`, `list`, `set`, and `dict`

How does the types
- `tuple`, `()`
- `list`, `[]`
- `set`, `{}`
- `dict`, `{}`

differ?

````python
t = (a**2 for a in range(-5, 5))
print(list(t))
````

````python
l = [a**2 for a in range(-5, 5)]
print(l)
````

````python
s = {a**2 for a in range(-5,5)}
print(s)
````

````python
d = {a : a**2 for a in range(-5, 5)}
print(d)
````

- How do you choose between `tuple` and `list`?

- When is a `set` useful?

- What makes `dict` different from the others?
