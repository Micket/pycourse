# General instructions for computer assignments<br> DAT171 (2019/2020)To receive the grade Passed on the computer assignments please follow the instructions below.
General

1. All groups work individually and write their own code etc. No code-sharing is allowed.
2.	In Python, classes are written in *CamelCase* where a capimtal letter is used for each new word, i.e. `ClassName`. Libraries, variables and functions are generally written in lowercase using underscores for word separation, i.e. `variable_name`.
3.	There are many more style recommendations for Python which can be read here: https://www.python.org/dev/peps/pep-0008/. If you are using PyCharm as an __IDE__ you will see a yellow line where you have broken the style guide. You do not have to follow all of the style guide, but if PyCharm has yellow stuff everywhere you probably want to fix some.
4.	Do not use global variables where it easily can be avoided.
5.	Constants should be defined in the beginning of the program (or function/class etc. when they are local).
6.	Constants are usually written in all caps, i.e. `BASE_RADIUS`. 
7.	Methods that change the state of the program should generally be written as a verb. For example, a method that removes a player from a game should be called `remove_player(player)` instead of `player_remove(player)`.
8.	It is almost always a bad idea to import things in the middle of a function. I.e. gather all your import statements at the top of the code unit they belong to.
9.	Keep your function declaration above the main code. I.e. the main code of your program should be at the end of the program file.
10.	Use the methods of an object directly, and not through the defining class.
11.	Try not to have unused variables in your program. *An unused variable hints that something is wrong*, did you forget to actually use the variable?
12.	As for Matlab, avoid using loops where it easily can be avoided. Use more efficient NumPy-arrays or similar. 

## Documentation
1.	Comment on lines or blocks of code to make the intent of it clear.
I.e. comment on *why* something is done or *where* expressions, equations, algorithms etc. stem from, __not__ that the line adds numbers `a` and `b` for `x = a + b` for example.
2.	Do not over comment, if the code is clear, comments are usually not needed on most lines!
